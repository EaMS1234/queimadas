```python
import pandas as pd
```

```python
df = pd.read_csv('bdqueimadas.csv')
```

```python
# Colunas não utilizadas pra treinamento
df = df.drop(columns=['Satelite', 'Pais', 'Estado', 'Municipio', 'FRP'])

# Apaga os dados dos biomas que não sejam mata atlântica e cerrado
df = df[df["Bioma"].isin(["Mata Atlântica", "Cerrado"])]

# Converte o bioma em um valor numérico baseado no tipo de vegetação.
# Esse valor é a constante A definida em: http://mtc-m21c.sid.inpe.br/col/sid.inpe.br/mtc-m21c/2019/11.21.11.03/doc/publicacao.pdf
dados = {'Mata Atlântica': 1.7, 'Cerrado': 3}
df['Bioma'] = df['Bioma'].map(dados)

# Filtra os dados quebrados
df = df[df["DiaSemChuva"] >= 0]
df = df[df["RiscoFogo"] >= 0]

# Arredonda a localização
df['Latitude'] = df['Latitude'].round(1)
df['Longitude'] = df['Longitude'].round(1)

# Remove a hora, e fica só com o dia
df['DataHora'] = pd.to_datetime(df['DataHora']).dt.date

# Filtra as entradas repetidas
df = df.drop_duplicates(subset=['DataHora', 'Latitude', 'Longitude'], keep='first')
```
